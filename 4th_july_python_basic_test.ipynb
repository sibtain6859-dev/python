```python
# 2. Write a program to determine whether a predefined number is divisible by both 5 and 7.

number = 35

if number % 5 == 0 and number % 7 == 0:
    print('divisible by both :',number)
else:
    print('not divisible',number)
```

```python
# 3. Write a program to find the largest among four predefined numbers using nested if statements.

a = 32
b = 40
c = 20
d = 35

if a >= b and a >= c and a >= d:
    print('is the largest', a)

elif b >= a and b >= c and b >= d:
    print('is the largest', b)

elif c >= a and c >= b and c >= d:
    print('is the largest', c)

else:
    print('is the largest', d)
```

```python
# 4. Write a grading program using if-elif-else: 90+=A, 75-89=B, 60-74=C, 40-59=D, below 40=Fail.

marks = int(input("Enter your marks: "))

if 90 <= marks <= 100:
    print('Grade: A')

elif 75 <= marks <= 89:
    print('Grade: B')

elif 60 <= marks <= 74:
    print('Grade: C')

elif 40 <= marks <= 59:
    print('Grade: D')

else:
    print('Fail')
```

```python
# 5. Write a program to check whether a year is a leap year using the complete leap year rules.

year = int(input('Enter a year : '))

if year % 4 == 0 and year % 100 != 0:
    print('Leap year')
elif year % 400 == 0:
    print('Leap year')
else:
    print('Not a leap year')
```

```python
# 6. Write a program to print all numbers from 1 to 50 that are divisible by 4 but not by 6.

for i in range(1,51):
  if i % 4 == 0 and i % 6 != 0:
    print(i)
```

```python
# 7. Write a program to print the squares of numbers from 1 to 15 using a for loop.

for number in range(1, 16):
    print(number * number)
```

```python
# 8. Write a program to calculate the factorial of a predefined number using a for loop.

number = 5
factorial = 1

for i in range(1, number + 1):
    factorial*=i

print(number, factorial)
```

```python
# 9. Write a program to count how many numbers between 1 and 100 are divisible by 3.

count = 0
for i in range(1,100):
  if i % 3 == 0:
    count+= 1
print(count)
```
