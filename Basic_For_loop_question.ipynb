```python
# 1. Write a program to print numbers from 1 to 10 using a for loop.

for i in range(1,11):
  print(i)
```

```python
# 2. Write a program to print all even numbers from 1 to 20 using a for loop.

for i in range(2,21):
  if i % 2 == 0:
    print(i)
```

```python
# 3. Write a program to print the multiplication table of 5 using a for loop.

for i in range(5,51):
  if i % 5 == 0:
    print(i)
```

```python
# 4. Write a program to print each character of the string 'PYTHON' on a separate line using a for loop.

name = 'PYTHON'

for i in name:
  print(i)
```

```python
# 5. Write a program to count how many vowels are present in the string 'programming' using a for
 #loop and if statement.

word = 'programming'
count = 0

for i in word:
    if i in 'aeiou':
        count+=1

print('Number of vowels:', count)



```

```python
# 6. Write a program to print numbers from 1 to 20 and display 'Even' or 'Odd' beside each number
# using a for loop and if-else.

for i in range(1,21):
  if i % 2 == 0:
    print('even number :',i)
  else:
    print('odd number  :',i)
```

```python
# 7. Write a program to find the sum of all numbers from 1 to 100 using a for loop.

sum = 0
for i in range(1,101):
  sum+=i
print(sum)

```
