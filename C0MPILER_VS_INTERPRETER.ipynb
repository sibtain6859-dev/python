```python
# COMPILER VS INTERPRETED LANGUAGE

'''
1. What is a Compiler?
A compiler is a software program that translates the entire source code into machine code (or another lower-level executable form) before execution.

In simple words:

A compiler first translates your program, and then the computer executes the translated program.

How does a compiler work?

Step 1: Source Code

Step 2: Compilation

The compiler translates your source code into machine code.

During compilation, it also checks for syntax and other compilation errors.

Step 3: Machine Code

The compiler generates machine code, usually stored in an object file before the linker creates the final executable.

Step 4: Execution

The operating system loads the executable, and the CPU executes its instructions.

Step 5: Output


    C Source Code
          |
          |
          v
       Compiler
          |
          |
          v
     Machine Code
          |
          |
          v
      Executable
          |
          |
          v
       CPU Runs
          |
          |
          v
        Output



What is an Interpreter?
Definition (Interview Answer)
An interpreter is a software program that processes and executes source code at runtime by translating or otherwise handling instructions as needed.
Meaning of Interpreter
The word interpreter means someone who explains or translates one language into another so that someone else can understand it.

For example:

Imagine two people:
- Person A speaks English.
- Person B speaks Japanese.
If Person A doesn't understand Japanese, an interpreter helps communicate between them.

Similarly, in programming, an interpreter helps the computer execute instructions written in a programming language.


How does an interpreter work?

Step 1: Source Code

You write your program in Python.

Step 2: Interpretation

The interpreter processes the instructions at runtime.

Step 3: Execution

The interpreter handles the execution of the program using the underlying computer system.

Step 4: Output


    Python Source Code
            |
            |
            v
        Interpreter
            |
            |
            v
         Execution
            |
            |
            v
          Output


What is the actual difference between them?

Imagine you have a book written in English, but you only understand Hindi.

Compiler approach:

You translate the entire book into Hindi first. After completing the translation, you read the translated book.

Interpreter approach:

Someone translates the English book for you as you read it, explaining each part as you go.

The difference is:
- Compiler: Translates the program before execution.
- Interpreter: Processes and executes the program at runtime.




'''
```

```python
# Q1. Why does a computer need a compiler or interpreter?

# Because a computer's CPU executes machine-code instructions.
# A compiler or interpreter helps process instructions written in high-level programming languages.
```

```python
# Q3. Which is faster, a compiler or an interpreter?

# Generally, compiled programs execute faster because their code has already been translated into machine code before execution.
# However, actual performance depends on the language, implementation and runtime environment.
```

```python
# Q4. Does an interpreter translate the entire program at once?

# Not necessarily. Traditional interpreters process instructions during execution.
# Different interpreter implementations work differently.
```

```python
# Q5. What is machine code?

# Machine code is a set of binary-encoded instructions that a computer's processor can execute directly.
```

```python
# Q2. Is Python a compiled or interpreted language?

# Python is commonly classified as an interpreted language.
# However, in CPython, source code is first compiled into bytecode, which is then executed by the Python Virtual Machine.


# Interview Tip: If the interviewer asks whether Python is compiled or interpreted, the best answer is:
# "Python is an interpreted language that internally uses compilation to convert source code into bytecode before execution."
```
