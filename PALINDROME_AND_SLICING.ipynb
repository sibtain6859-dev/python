```python
# Palindrome type question on integer

n = int(input('Enter a number :'))

original = n
reverse = 0

while (n > 0):
  digit = n % 10
  reverse = reverse * 10 + digit
  n = n // 10
if original == reverse:
  print('Palindrome')
else:
  print('Not palindrome')

```

```python
n = input('Enter string :')

original = n
reverse = ''
i = len(n) -1

while ( i >= 0):
  reverse = reverse + n[i]
  i = i - 1

if original == reverse:
  print('Palindrome string')
else:
  print('Not a Palindrome string')
```
