```python
# Pattern question

n = int(input('Enter a number :'))
print('*'*n)
```

```python
n = int(input('Enter a number :'))

for i in range(n):
  print('*')
```

```python
n = int(input('Enter a number :'))

for i in range(n):
  print('* ',end='')                  # end=' '  tells python to add space instead of moving to the next line
                                      # \n  means new line
```

```python
n = int(input('Enter a number of rows :'))

for i in range(1, n + 1 ):

  for j in range(1, i + 1):
    print('* ', end='')
  print()
```

```python
# inverted right angle triangle

n = int(input('Enter the number of rows :'))

for i in range(n, 0, -1):

  for j in range(0, i):
    print('* ',end='')
  print()
```

```python
n = int(input('Enter the number of rows :'))
k = 1

for i in range(1, n + 1):

  for j in range(1, k + 1):
    print('* ',end='')

  k = k +  2
  print()
```

```python
# pyramid like structure

n = int(input('Enter the number of rows :'))

for i in range(0, n):

  for j in range(0, n -i-1):
    print(' ',end='')

  for j in range(0, i + 1):
    print(' * ',end='')
  print()
```

```python
# Reverse pyramid

n = int(input('Enter the number of rows :'))

for i in range(n, 0 , -1):

  for j in range(0, n -i):
    print(' ',end='')

  for j in range(0,i):
    print('* ',end='')

  print()
```

```python
# Diamond like structure

n = int(input('Enter the number of rows :'))

# upper  half
for i in range(0,n):
  for j in range(0, n-i-1):
    print(' ',end='')

  for j in range(0, 2*i+1):
    print('*',end='')

  print()

# lower half
for i in range(n-2, -1, -1):
  for j in range(0, n-i-1):
    print(' ',end='')

  for j in range(0, 2*i+1):
    print('*',end='')

  print()

```
